#**Q1. What is a loop, and why do we use loops in programming?**
* **Definition**: A loop is a programming construct that repeats a block of code multiple times until a specific condition is met.  
* **Purpose**: We use loops to automate repetitive tasks, reduce code redundancy, save time, and make programs cleaner, more efficient, and easier to maintain.

# **2. Explain the difference between for and while loops.**
* **For Loop:** A `for` loop is typically used when the number of iterations is known beforehand; it iterates over a sequence such as a list, tuple, or range.  
* **While Loop:** A `while` loop is used when the number of iterations is not known in advance and depends on a boolean condition that is evaluated before each iteration.

# **Q3. What is the purpose of the range function?**
* **Purpose**: The `range()` function in Python is used to generate a sequence of numbers, commonly used with `for` loops to iterate a specific number of times. It can take parameters for start, stop, and step values.

# **Q4 Write a program to print numbers from 1 to 100.**

```python
for i in range(1, 101):
  print(i)
```

# **Q5. Write a program to print all even numbers within a user-given range.**

```python
start = int(input("Enter the start of the range: "))
end = int(input("Enter the end of the range: "))

print(f"Even numbers between {start} and {end}:")
for i in range(start, end + 1):
    if i % 2 == 0:
        print(i)
```

#**Q6. Write a program to display the multiplication table of a given number.**

```python
num = int(input("Enter a number: "))
print(f"Multiplication Table for {num}:")
for i in range(1, 11):
    print(f"{num} x {i} = {num * i}")
```

# **Q7. Write a program to calculate the factorial of a number.**

```python
num = int(input("Enter a number: "))
factorial = 1

if num < 0:
    print("Factorial does not exist for negative numbers.")
else:
    for i in range(1, num + 1):
        factorial *= i
    print(f"The factorial of {num} is {factorial}")
```

#**Q8. Write a program to generate the first ten terms of the Fibonacci sequence.**

```python
n_terms = 10
n1, n2 = 0, 1
count = 0

print("Fibonacci sequence:")
while count < n_terms:
    print(n1, end=" ")
    nth = n1 + n2
    n1 = n2
    n2 = nth
    count += 1
```

#**Q9. Write a program to check whether a number is prime.**

```python
num = int(input("Enter a number: "))

if num <= 1:
    print(f"{num} is not a prime number.")
else:
    is_prime = True
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            is_prime = False
            break

    if is_prime:
        print(f"{num} is a prime number.")
    else:
        print(f"{num} is not a prime number.")
```

#**Q10. Write a program to count the digits in an integer.**

```python
num = int(input("Enter an integer: "))
count = 0
temp = abs(num)

if temp == 0:
    count = 1
else:
    while temp > 0:
        count += 1
        temp //= 10

print(f"The number of digits in {num} is {count}.")
```
