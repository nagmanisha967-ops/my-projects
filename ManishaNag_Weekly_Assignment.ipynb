# **Q1. What is a list? Mention three important characteristics of lists.**
* Answer:

* **Defination:**  A list is a built-in data structure in Python used to store a collection of items in a single variable. They are created using square brackets [].

**Three important characteristics:**


1.  **Ordered:** The items maintain a defined order, and that order doesn't change unless explicitly modified.

2.  **Mutable**: You can change, add, or remove items
after the list has been created.

3. **Allow duplicates :** Lists can contain items with the exact same value.

# **Q2. What is the differenc between append() and extend()?**
* Answer:
* **`append()`:** Adds its argument as a single element to the end of the list. If you append a list to another list, the whole list is added as a nested sub-list.

* **`extend()`:** Iterates over its argument and adds each element individually to the end of the list, effectively flattening iterables like lists or tuples into the main list.

# **Q3. Write a program to find the largest and smallest values in a list.**
* Answer.


```python
numbers = [45, 12, 89, 23, 3, 67]

largest = max(numbers)
smallest = min(numbers)

print(f"The largest value is: {largest}")
print(f"The smallest value is: {smallest}")
```

# **Q4. Write a program to calculate the sum and average of numbers stored in a list.**
* Answer:

```python
numbers = [10, 20, 30, 40, 50]

total_sum = sum(numbers)
average = total_sum / len(numbers) if numbers else 0

print(f"Sum: {total_sum}")
print(f"Average: {average}")
```

# **Q5. Write a program to remove duplicate values from a list.**
* Answer:

```python
original_list = [1, 2, 2, 3, 4, 4, 5]

# Convert to a set to remove duplicates, then back to a list
unique_list = list(set(original_list))

print(f"List without duplicates: {unique_list}")
```

# **Q6. What is a tuple? How is it different from a list?**
* Answer:
* **Defination:** A tuple is a collection data structure in Python used to store multiple items in a single variable, defined using parentheses ().
* **Difference from a list:** The main difference is that tuples are immutable (cannot be modified, added to, or changed after creation), whereas lists are mutable. Because of this, tuples are generally faster and use less memory.

# **Q8. What is a set? Why are sets useful for removing duplicates?**
* Answer:
* **Defination:**  A set is an unordered, unindexed collection of unique items enclosed in curly brackets`{}`

* **Why it's useful:**  By definition, sets cannot contain duplicate elements. When you pass a list with duplicate values into the `set()` constructor, Python automatically filters out all repeating elements instantly.

# **Q9. Write a program to find common elements between tow collections.**
* Answer:

```python
set_a = {1, 2, 3, 4, 5}
set_b = {4, 5, 6, 7, 8}

# Using set intersection
common_elements = set_a.intersection(set_b)

print(f"Common elements: {common_elements}")
```

# **Q10. What is a dictionary? Explain keys and values.**
* **Defination:** A dictionary is a collection used to store data in key-value pairs, enclosed in curly brackets {} with colon separators.
* **Keys and Values:**

* **Keys:**  Unique identifiers used to look up data. They must be immutable types (like strings, numbers, or tuples).
Values: The actual data associated with those keys, which can be of any data type and can contain duplicates.
* **Example:** `{"name": "Alice", "age": 25} `where `"name"` and `"age"` are keys, and `"Alice"` and `25` are values.
