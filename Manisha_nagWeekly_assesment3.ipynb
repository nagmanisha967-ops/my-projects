# **Q1. What is the purpose of an if statement?**

* The purpose of an `if` statement in Python is decision-making. It executes a block of code only if a specified condition evaluates to `True`. If the condition is `False`, the code block inside the `if` statement is skipped.

#  **Q2. Explain the difference between if, if-else, and if-elif-else.**

* `if` statement: Evaluates a single condition. The code inside executes only if the condition is `True`.

*` `if-else` `statement: Provides two paths. If the condition is `True`, the `if` block executes; otherwise, the `else` block executes.

* `if-elif-else` statement: Used to test multiple conditions sequentially. If the first condition is `False`, it checks the `elif` conditions one by one. If none are `True`, the final `else` block executes.

# **Q3. Write a program to check whether a number is positive, negative or zero.**

```python
number = float(input("Enter a number: "))

if number > 0:
    print("The number is Positive.")
elif number < 0:
    print("The number is Negative.")
else:
    print("The number is Zero.")
```

#**Q4. Write a program to check whether a number is even or odd.**

```python
number = int(input("Enter an integer: "))

if number % 2 == 0:
    print("The number is Even.")
else:
    print("The number is Odd.")
```


#**Q5. Write a program to find the greatest among three numbers.**

```python
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
num3 = float(input("Enter third number: "))

if num1 >= num2 and num1 >= num3:
    greatest = num1
elif num2 >= num1 and num2 >= num3:
    greatest = num2
else:
    greatest = num3

print(f"The greatest number is {greatest}")
```

#**Q6. Write a program to check whether a given year is a leap year.**

```python
year = int(input("Enter a year: "))

if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
    print(f"{year} is a Leap Year.")
else:
    print(f"{year} is not a Leap Year.")
```

**Q7. Write a program to assign grades based on a student's marks.**

```python
marks = float(input("Enter student marks (0-100): "))

if marks >= 90:
    grade = 'A+'
elif marks >= 80:
    grade = 'A'
elif marks >= 70:
    grade = 'B'
elif marks >= 60:
    grade = 'C'
elif marks >= 50:
    grade = 'D'
else:
    grade = 'F (Fail)'

print(f"Grade: {grade}")
```

#**Q9. Write a program to calculate an electricity bill using different usage slabs.**

```python
units = float(input("Enter electricity units consumed: "))
bill_amount = 0

if units <= 100:
    bill_amount = units * 5
elif units <= 200:
    bill_amount = (100 * 5) + ((units - 100) * 7)
elif units <= 300:
    bill_amount = (100 * 5) + (100 * 7) + ((units - 200) * 10)
else:
    bill_amount = (100 * 5) + (100 * 7) + (100 * 10) + ((units - 300) * 12)

print(f"Total Electricity Bill: ₹{bill_amount:.2f}")
```

#**Q10. What is a nested if statement? Give a suitable situation where it can be used.**

* A nested if statement is an if (or if-else) block placed inside another if or else block. It allows you to test multiple conditions that depend on one another.  
* **Suitable Situation:**

Login authentication where you first verify if the username exists. Only if the username is correct do you check if the password matches.

```python
# Example of Nested If Statement
username = input("Enter username: ")
password = input("Enter password: ")

if username == "admin":
    if password == "12345":
        print("Login successful!")
    else:
        print("Incorrect password.")
else:
    print("User not found.")
```
